```python
import numpy as np
import matplotlib.pyplot as plt
def trapz(xn,yn):
    N = len(xn) #how many nodeal areas
    A=0
    for n in range(0,N-1): # one trapezia less than nodes

        A += 0.5*(yn[n] + yn[n+1] ) * ( xn[n+1] - xn[n] )
    return A #area
```

```python
xn=np.linspace(0,2*np.pi,50)
yn=np.sin(xn)
plt.scatter(xn,yn)
A=trapz(xn,yn)
print(A)
#accurqcy vs computational time tradeoff
```
