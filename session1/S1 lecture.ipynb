```python
import mpl_toolkits as mplot3d # redundant from matplotlib

dx=0.1
x=np.arange(1,7+dx,dx)
dy=0.2
y=np.arange(1,5+dy,dy) #using two one dimensional array = Meshing
Nx = len(x)
Ny = len(y)
T=np.ndarray((Ny,Nx))
#T=x**2+y**2  -- error mismatch of dimensions 5x7 not equal 1x5 + 1x5
#can use looping or Grid method
Xg , Yg = np.meshgrid(x,y)
T=Xg**2+Yg**2 #computationally same as loop, but same as maths, shorter to write, elegant
print(T)

ax = plt.axes(projection='3d')
ax.plot_surface(Xg,Yg,T)
#ax.plot_wireframe(Xg,Yg,T) # alternative
plt.show() #new plot
plt.contour(Xg,Yg,T)
#plt.contour(x,y,T) # also works
```
