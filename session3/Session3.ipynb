```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def Lagrangian(j,xn,xp):
    N=len(xn)
    L=1
    for k in range(N):
        if k!=j:
            L=L*(xp-xn[k])/(xn[j]-xn[k])
    return L
```

```python
def LagInt(xn,yn,xp):
    yp=0
    N=len(xn)
    for j in range(N):
        yp+=yn[j]*Lagrangian(j,xn,xp)
    return yp
```

```python
xn = np.linspace(0,2*np.pi,5)
yn=np.sin(xn)

x=np.linspace(0,2*np.pi,20)
y=[]
for xp in x:
    yp = LagInt(xn,yn,xp)
    y.append(yp)

y=np.array(y)
plt.scatter(xn,yn,c='r')
plt.scatter(x,y,c='b')
plt.grid()
```

```python
#newton's divided difference method
def Newton(xn,yn):
    n=len(xn)
    if n==1:
        y=yn[0]
    else:
        y= (Newton(xn[0:n-1],yn[0:n-1]) - Newton(xn[1:],yn[1:]) )/  (xn[0]-xn[n-1])
    return y
```
