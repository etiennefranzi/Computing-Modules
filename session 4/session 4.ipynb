```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy.special as sp
```

```python
def Simpson(xn,yn):
    
    N=len(xn) #number of nodes
    h= xn[1]-xn[0]
    I=0
    for i in range(0,N-2,2): #0 is important here
        I+=h*1/3*(yn[i] + 4*yn[i+1] + yn[i+2] )
    
    return I
```

```python
xn=np.linspace(0,2*np.pi,1000)
yn=np.sin(xn)
print(Simpson(xn,yn))
```

```python
def SimpsonCompound(h,yn): # as explained in Maths class, does not take x
    N = len(yn) # number of nodal points
    # check that there is an even number of intervals
    if (N-1) % 2 != 0: #(N-1 is the number of intervals)
        print('warning; there are ',N-1,' intervals')
    M = int((N-1)/2) # half number of interval, ignores last interval
    S = yn[0] + yn[2*M]  # initialise sum
    for i in range(1,2*M,2):
        S += 4*yn[i]
    for i in range(2,2*M-1,2):
        S += 2*yn[i]
    S = S * h / 3
    
    return S
```

```python
import numpy as np
import scipy.special as sp

def func(x): # this is the function we wish to integrate
    f = np.sin(x)
    return f

def Gaussian(n):
    xn, wn = sp.roots_legendre(n)
    I = 0
    for i in range(n):
        I += wn[i] * func(xn[i])
    return I
```
